```python
# =========================================================
# 🩸 BLOOD CELL DETECTION — CLEAN TRAINING PIPELINE v9
# Merges BCCD + Custom dataset, trains properly (no hacky filters)
# =========================================================

!pip install ultralytics -q

import os, shutil, zipfile, glob, yaml
from pathlib import Path
from google.colab import files
from ultralytics import YOLO

print("✅ Setup done!")

# =========================================================
# STEP 1: UPLOAD BOTH DATASET ZIPS
# =========================================================
print("\n📤 Upload BCCD dataset ZIP (YOLOv8 format)")
bccd_upload = files.upload()
bccd_zip = list(bccd_upload.keys())[0]

print("\n📤 Upload YOUR custom dataset ZIP (YOLOv8 format)")
custom_upload = files.upload()
custom_zip = list(custom_upload.keys())[0]

# =========================================================
# STEP 2: EXTRACT BOTH
# =========================================================
for folder in ['/content/bccd_raw', '/content/custom_raw', '/content/merged']:
    if os.path.exists(folder):
        shutil.rmtree(folder)
    os.makedirs(folder, exist_ok=True)

with zipfile.ZipFile(bccd_zip, 'r') as z:
    z.extractall('/content/bccd_raw')
with zipfile.ZipFile(custom_zip, 'r') as z:
    z.extractall('/content/custom_raw')

print("✅ Both datasets extracted!")

# =========================================================
# STEP 3: MERGE INTO ONE DATASET (train/valid/test combined)
# =========================================================
merged_root = '/content/merged'
for split in ['train', 'valid', 'test']:
    os.makedirs(f'{merged_root}/{split}/images', exist_ok=True)
    os.makedirs(f'{merged_root}/{split}/labels', exist_ok=True)

def copy_split(src_root, prefix, split):
    img_src = os.path.join(src_root, split, 'images')
    lbl_src = os.path.join(src_root, split, 'labels')
    if not os.path.exists(img_src):
        return 0
    count = 0
    for img_file in glob.glob(f'{img_src}/*'):
        name = Path(img_file).stem
        ext = Path(img_file).suffix
        new_name = f'{prefix}_{name}'
        shutil.copy(img_file, f'{merged_root}/{split}/images/{new_name}{ext}')
        lbl_file = os.path.join(lbl_src, name + '.txt')
        if os.path.exists(lbl_file):
            shutil.copy(lbl_file, f'{merged_root}/{split}/labels/{new_name}.txt')
        count += 1
    return count

total = 0
for split in ['train', 'valid', 'test']:
    n1 = copy_split('/content/bccd_raw', 'bccd', split)
    n2 = copy_split('/content/custom_raw', 'custom', split)
    print(f"  {split}: BCCD={n1}, Custom={n2}, Total={n1+n2}")
    total += n1 + n2

print(f"\n✅ Merged dataset: {total} total images")

# =========================================================
# STEP 4: CREATE COMBINED data.yaml
# =========================================================
data_yaml = {
    'path': merged_root,
    'train': 'train/images',
    'val': 'valid/images',
    'test': 'test/images',
    'nc': 3,
    'names': ['Platelets', 'RBC', 'WBC']
}
with open(f'{merged_root}/data.yaml', 'w') as f:
    yaml.dump(data_yaml, f)

print("✅ Combined data.yaml created!")
print(open(f'{merged_root}/data.yaml').read())
```

```python
# =========================================================
# STEP 5: TRAIN — full epochs, no manual hacks
# =========================================================
model = YOLO('yolo11n.pt')   # fresh pretrained COCO weights se start

results = model.train(
    data=f'{merged_root}/data.yaml',
    epochs=100,
    patience=20,          # early stop only if truly no improvement
    imgsz=640,
    batch=16,
    device=0,              # GPU (Colab: Runtime → Change runtime type → GPU rakhna zaroori)
    project='/content/runs',
    name='blood_merged_v9',
    optimizer='auto',
    lr0=0.001,
    cos_lr=True,
    augment=True,           # built-in YOLO augmentation, hardcoded rules nahi
    plots=True,
    verbose=True
)

print("\n✅ TRAINING COMPLETE!")

# Best model path
best_model_path = glob.glob('/content/runs/blood_merged_v9*/weights/best.pt')[-1]
print(f"📍 Best model saved at: {best_model_path}")

# Print final validation metrics
metrics = model.val(data=f'{merged_root}/data.yaml')
print("\n📊 FINAL METRICS:")
print(f"  Precision: {metrics.box.mp:.3f}")
print(f"  Recall: {metrics.box.mr:.3f}")
print(f"  mAP50: {metrics.box.map50:.3f}")
print(f"  mAP50-95: {metrics.box.map:.3f}")

files.download(best_model_path)
print("📥 best.pt downloaded!")
```

```python
# =========================================================
# STEP 6: TEST ON A NEW IMAGE — clean, no hardcoded filters
# =========================================================
print("\n📤 Upload a test blood smear image")
test_upload = files.upload()
test_img = list(test_upload.keys())[0]

detect_model = YOLO(best_model_path)

results = detect_model.predict(
    source=test_img,
    conf=0.35,          # reasonable confidence threshold (not hardcoded post-hoc filtering)
    iou=0.5,
    imgsz=640,
    save=True,
    show_labels=True,
    show_conf=True
)

r = results[0]
class_counts = {'Platelets': 0, 'RBC': 0, 'WBC': 0}
names = r.names
for cls_id in r.boxes.cls.cpu().numpy():
    class_counts[names[int(cls_id)]] += 1

print("\n🩸 DETECTION RESULTS")
print(f"  RBC: {class_counts['RBC']}")
print(f"  WBC: {class_counts['WBC']}")
print(f"  Platelets: {class_counts['Platelets']}")
print(f"  Avg Confidence: {r.boxes.conf.mean().item():.3f}")

# Show result image
r.show()
r.save(filename='/content/result_v9.jpg')
files.download('/content/result_v9.jpg')
```

```python
# =========================================================
# STEP 7: MULTI-DISEASE RISK SCREENING
# Rule-based weighted scoring — explainable, transparent
# Diseases: Anemia | Sickle Cell | Thalassemia | Leukemia
# =========================================================

import numpy as np
import matplotlib.pyplot as plt

rbc_count = class_counts['RBC']
wbc_count = class_counts['WBC']
plat_count = class_counts['Platelets']

# --- Reference midpoints (tune to your imaging setup) ---
RBC_MID, RBC_LOW, RBC_HIGH = 225, 150, 300
WBC_MID, WBC_LOW, WBC_HIGH = 8, 4, 12
PLAT_MID, PLAT_LOW, PLAT_HIGH = 32, 20, 45

# Normalized deviation scores (0 = normal, 1 = far from normal)
def deviation_score(value, low, high):
    if low <= value <= high:
        return 0.0
    elif value < low:
        return min(1.0, (low - value) / low)
    else:
        return min(1.0, (value - high) / high)

rbc_dev = deviation_score(rbc_count, RBC_LOW, RBC_HIGH)
wbc_dev = deviation_score(wbc_count, WBC_LOW, WBC_HIGH)
plat_dev = deviation_score(plat_count, PLAT_LOW, PLAT_HIGH)

rbc_low_flag  = 1.0 if rbc_count < RBC_LOW else 0.0
wbc_high_flag = 1.0 if wbc_count > WBC_HIGH else 0.0
wbc_low_flag  = 1.0 if wbc_count < WBC_LOW else 0.0
plat_low_flag = 1.0 if plat_count < PLAT_LOW else 0.0

# --- Weighted rule-based scoring per disease ---
# Weights reflect known clinical association patterns (simplified heuristic)
disease_scores = {
    "Anemia": (
        0.7 * rbc_dev * rbc_low_flag +
        0.3 * plat_dev
    ),
    "Sickle Cell Anemia": (
        0.6 * rbc_dev * rbc_low_flag +
        0.4 * plat_dev * plat_low_flag
    ),
    "Thalassemia": (
        0.5 * rbc_dev +
        0.3 * (1 - wbc_dev) +   # thalassemia often has near-normal WBC
        0.2 * plat_dev
    ),
    "Leukemia": (
        0.6 * wbc_dev * wbc_high_flag +
        0.4 * plat_dev * plat_low_flag
    )
}

# Convert raw scores to percentages (softmax-style normalization for readability)
raw = np.array(list(disease_scores.values()))
raw = np.clip(raw, 0, 1)

# Add small baseline so "normal" case doesn't force weird 25% each
if raw.sum() == 0:
    percentages = np.zeros(4)
else:
    exp_scores = np.exp(raw * 3)  # temperature scaling
    percentages = (exp_scores / exp_scores.sum()) * 100 if raw.sum() > 0 else exp_scores*0

disease_names = list(disease_scores.keys())

print("=" * 55)
print("🩺 DISEASE RISK SCREENING (Rule-Based, Indicative)")
print("=" * 55)
print(f"  RBC: {rbc_count}  |  WBC: {wbc_count}  |  Platelets: {plat_count}")
print()
print("  Risk Likelihood Scores:")
overall_normal = (rbc_dev + wbc_dev + plat_dev) == 0
for name, pct in zip(disease_names, percentages):
    bar = "█" * int(pct/5)
    print(f"    {name:20s}: {pct:5.1f}%  {bar}")

if overall_normal:
    print("\n  ✅ All counts within normal range — LOW overall risk")
else:
    top_disease = disease_names[np.argmax(percentages)]
    print(f"\n  ➡ Highest indicative risk: {top_disease} ({max(percentages):.1f}%)")

print()
print("=" * 55)
print("⚠️  These are RULE-BASED indicative scores derived from")
print("   cell-count deviations, NOT a validated diagnostic")
print("   probability. Purpose: flag samples for priority")
print("   clinical follow-up in resource-limited settings.")
print("   Always confirm with a certified pathologist / CBC test.")
print("=" * 55)

# --- Visualization ---
fig, ax = plt.subplots(figsize=(9, 5))
colors = ['#e63946' if p == max(percentages) and not overall_normal else '#457b9d' for p in percentages]
bars = ax.barh(disease_names, percentages, color=colors)
ax.set_xlabel('Risk Likelihood Score (%)')
ax.set_title('Disease Risk Screening — Indicative Only')
ax.set_xlim(0, 100)
for bar, pct in zip(bars, percentages):
    ax.text(pct + 1, bar.get_y() + bar.get_height()/2, f'{pct:.1f}%', va='center')
plt.figtext(0.5, -0.02, "⚠ Rule-based screening indicator — not a clinical diagnosis. Confirm with a doctor.",
            ha='center', fontsize=9, style='italic', color='gray')
plt.tight_layout()
plt.savefig('/content/disease_risk_result.png', dpi=150, bbox_inches='tight')
plt.show()
files.download('/content/disease_risk_result.png')
```
